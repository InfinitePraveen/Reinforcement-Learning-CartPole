# 01 - CartPole Environment Setup
Understand the environment, observation space, action space and rewards.

```python
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
```

```python
env = gym.make('CartPole-v1')
print('Observation space:', env.observation_space)
print('Action space:', env.action_space)
```

```python
observation, info = env.reset(seed=42)
print('Initial observation:', observation)
```

```python
action = env.action_space.sample()
observation, reward, terminated, truncated, info = env.step(action)
print('Action:', action)
print('Next observation:', observation)
print('Reward:', reward)
print('Done:', terminated or truncated)
```

```python
print('Observation values:', observation)
print('Cart position:', observation[0])
print('Cart velocity:', observation[1])
print('Pole angle:', observation[2])
print('Pole angular velocity:', observation[3])
```

## Why discretization?
Q-learning stores a value for each state-action pair. CartPole observations are continuous, so we group continuous values into a small number of bins.

```python
bins = (8, 8, 12, 12)
print('Discrete state shape:', bins)
print('Number of discrete states:', np.prod(bins))
```

```python
state_limits = np.array([2.4, 3.0, 0.2095, 3.5], dtype=np.float32)
print('Approximate clipping limits:', state_limits)
```

```python
def discretize(observation):
    clipped = np.clip(observation, -state_limits, state_limits)
    scaled = (clipped + state_limits) / (2 * state_limits)
    indices = (scaled * np.array(bins)).astype(int)
    return tuple(np.clip(indices, 0, np.array(bins)-1))
```

```python
state = discretize(observation)
print('Discrete state:', state)
```
